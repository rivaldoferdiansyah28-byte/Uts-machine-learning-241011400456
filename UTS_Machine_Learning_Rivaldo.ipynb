## Bagian A – Linear Regression
Dataset dummy penjualan berdasarkan biaya iklan (40 baris data).

### Soal A1 – Load & Visualisasi Data

```python
# Membuat data dummy sesuai spesifikasi (40 baris, tren linier positif)
# Biaya Iklan (X) dalam Juta Rupiah, Penjualan (y) dalam Juta Rupiah
biaya_iklan = np.array([
    0.8, 0.9, 1.0, 1.3, 1.7, 2.0, 2.2, 2.5, 2.8, 3.0,
    3.2, 3.5, 3.8, 4.0, 4.3, 4.5, 4.8, 5.0, 5.3, 5.5,
    5.8, 6.0, 6.2, 6.5, 6.8, 7.0, 7.3, 7.5, 7.8, 8.0,
    8.2, 8.5, 8.8, 9.0, 9.2, 9.5, 9.7, 9.9, 8.3, 8.6
])

penjualan = np.array([
    34.3, 38.8, 33.2, 36.0, 51.3, 47.5, 49.0, 52.1, 55.4, 58.0,
    59.2, 62.5, 64.0, 67.2, 68.5, 70.1, 71.8, 74.5, 76.2, 78.0,
    80.1, 82.5, 83.9, 87.1, 89.0, 91.5, 93.2, 95.0, 98.2, 100.1,
    101.5, 105.0, 107.2, 109.5, 111.0, 114.2, 116.0, 118.5, 102.0, 104.5
])

print(f"Jumlah baris data: {len(biaya_iklan)}")
print("5 data pertama:")
for i in range(5):
    print(f"  * BiayaIklan = {biaya_iklan[i]}, Penjualan = {penjualan[i]}")

# Visualisasi Data
plt.figure(figsize=(8, 5))
plt.scatter(biaya_iklan, penjualan, color='blue', alpha=0.7, edgecolors='k')
plt.title('Scatter Plot: Biaya Iklan vs Penjualan')
plt.xlabel('Biaya Iklan (Juta Rupiah)')
plt.ylabel('Penjualan (Juta Rupiah)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Soal A2 – Split Data

```python
# Fungsi Train-Test Split from Scratch dengan random_state=56
def train_test_split_custom(X, y, test_size=0.2, random_state=None):
    if random_state is not None:
        np.random.seed(random_state)
    
    n_samples = len(X)
    indices = np.arange(n_samples)
    np.random.shuffle(indices)
    
    n_test = int(n_samples * test_size)
    test_idx = indices[:n_test]
    train_idx = indices[n_test:]
    
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

X = biaya_iklan.reshape(-1, 1)
y = penjualan

X_train, X_test, y_train, y_test = train_test_split_custom(X, y, test_size=0.2, random_state=56)

print(f"Shape Dataset:")
print(f"  X_train: {X_train.shape} | X_test: {X_test.shape}")
print(f"  y_train: {y_train.shape} | y_test: {y_test.shape}")
```

### Soal A3 – Training dengan Normal Equation

```python
# Menambahkan kolom bias (intercept) 1 pada X_train
X_train_b = np.c_[np.ones((X_train.shape[0], 1)), X_train]

# Normal Equation: theta = (X^T * X)^(-1) * X^T * y
theta_best = np.linalg.inv(X_train_b.T.dot(X_train_b)).dot(X_train_b.T).dot(y_train)

intercept = theta_best[0]
coef = theta_best[1]

print(f"Intercept (theta_0): {intercept:.4f}")
print(f"Koefisien (theta_1): {coef:.4f}")
print(f"Persamaan Garis Regresi: Penjualan = {intercept:.4f} + {coef:.4f} * BiayaIklan")
```

### Soal A4 – Evaluasi Model

```python
# Fungsi Evaluasi Metrik
def evaluate_metrics(y_true, y_pred):
    mae = np.mean(np.abs(y_true - y_pred))
    mse = np.mean((y_true - y_pred) ** 2)
    rmse = np.sqrt(mse)
    ss_total = np.sum((y_true - np.mean(y_true)) ** 2)
    ss_res = np.sum((y_true - y_pred) ** 2)
    r2 = 1 - (ss_res / ss_total)
    return mae, mse, rmse, r2

# Prediksi pada train dan test
X_test_b = np.c_[np.ones((X_test.shape[0], 1)), X_test]
y_train_pred = X_train_b.dot(theta_best)
y_test_pred = X_test_b.dot(theta_best)

mae_train, mse_train, rmse_train, r2_train = evaluate_metrics(y_train, y_train_pred)
mae_test, mse_test, rmse_test, r2_test = evaluate_metrics(y_test, y_test_pred)

print("Tabel Evaluasi Model:")
print(f"{'Metrik Evaluasi':<15} | {'Data Training':<15} | {'Data Testing':<15}")
print("-" * 51)
print(f"{'MAE':<15} | {mae_train:<15.4f} | {mae_test:<15.4f}")
print(f"{'MSE':<15} | {mse_train:<15.4f} | {mse_test:<15.4f}")
print(f"{'RMSE':<15} | {rmse_train:<15.4f} | {rmse_test:<15.4f}")
print(f"{'R2 Score':<15} | {r2_train:<15.4f} | {r2_test:<15.4f}")
```

### Soal A5 – Gradient Descent & Learning Rate

```python
def gradient_descent(X, y, lr=0.02, n_iterations=5000):
    m = len(y)
    theta = np.zeros(2)
    X_b = np.c_[np.ones((m, 1)), X]
    
    for iteration in range(n_iterations):
        gradients = (2/m) * X_b.T.dot(X_b.dot(theta) - y)
        theta -= lr * gradients
    return theta

# Eksperimen dengan berbagai learning rate
learning_rates = [0.001, 0.02, 0.1]
for lr in learning_rates:
    try:
        theta_gd = gradient_descent(X_train, y_train, lr=lr, n_iterations=5000)
        print(f"LR = {lr:<6} -> Intercept: {theta_gd[0]:.4f}, Coef: {theta_gd[1]:.4f}")
    except Exception as e:
        print(f"LR = {lr:<6} -> Error: {e}")
```

### Soal A6 – Visualisasi & Prediksi

```python
# Prediksi untuk Biaya Iklan baru: 4.5, 7.0, dan 12.0 juta Rupiah
biaya_baru = np.array([4.5, 7.0, 12.0])
biaya_baru_b = np.c_[np.ones((len(biaya_baru), 1)), biaya_baru]
prediksi_penjualan = biaya_baru_b.dot(theta_best)

for b, p in zip(biaya_baru, prediksi_penjualan):
    print(f"Biaya Iklan {b} juta -> Prediksi Penjualan: {p:.4f} juta")

# Plot Garis Regresi
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Data Aktual', alpha=0.7)
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
x_line_b = np.c_[np.ones((len(x_line), 1)), x_line]
y_line = x_line_b.dot(theta_best)
plt.plot(x_line, y_line, color='red', linewidth=2, label='Garis Regresi')
plt.scatter(biaya_baru, prediksi_penjualan, color='green', s=100, zorder=5, label='Titik Prediksi Baru')
plt.title('Linear Regression Fit & Prediksi Baru')
plt.xlabel('Biaya Iklan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Bagian B – Logistic Regression
Dataset klasifikasi email spam berdasarkan Jumlah Link dan Kata Promo (60 baris data).

### Soal B1 – Load & Visualisasi Data

```python
# Data Dummy Email Spam (60 baris: [JumlahLink, KataPromo, LabelSpam])
np.random.seed(56)
n_data = 60
jumlah_link = np.random.randint(0, 10, size=n_data)
kata_promo = np.random.randint(0, 10, size=n_data)

# Buat label berdasarkan aturan logis + sedikit noise
label_spam = np.where((jumlah_link + kata_promo) > 10, 1, 0)

print(f"Total data: {n_data}")
print(f"Distribusi Target:")
print(f"  Bukan Spam (0): {np.sum(label_spam == 0)}")
print(f"  Spam (1): {np.sum(label_spam == 1)}")

# Visualisasi
plt.figure(figsize=(8, 5))
plt.scatter(jumlah_link[label_spam == 0], kata_promo[label_spam == 0], color='blue', label='Bukan Spam (0)', alpha=0.8, edgecolor='k')
plt.scatter(jumlah_link[label_spam == 1], kata_promo[label_spam == 1], color='red', label='Spam (1)', alpha=0.8, edgecolor='k')
plt.title('Scatter Plot: Jumlah Link vs Kata Promo')
plt.xlabel('Jumlah Link')
plt.ylabel('Kata Promo')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Soal B2 – Split Data & Training

```python
X_logis = np.column_stack((jumlah_link, kata_promo))
y_logis = label_spam

X_train_l, X_test_l, y_train_l, y_test_l = train_test_split_custom(X_logis, y_logis, test_size=0.25, random_state=56)

def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500)))

def train_logistic_regression(X, y, lr=0.1, n_iterations=10000):
    m, n = X.shape
    theta = np.zeros(n + 1)
    X_b = np.c_[np.ones((m, 1)), X]
    
    loss_history = []
    for i in range(n_iterations):
        z = X_b.dot(theta)
        h = sigmoid(z)
        gradient = X_b.T.dot(h - y) / m
        theta -= lr * gradient
        
        # Binary Cross-Entropy Loss
        loss = -np.mean(y * np.log(h + 1e-15) + (1 - y) * np.log(1 - h + 1e-15))
        loss_history.append(loss)
        
    return theta, loss_history

theta_logis, loss_hist = train_logistic_regression(X_train_l, y_train_l, lr=0.1, n_iterations=10000)

print(f"Intercept (theta_0): {theta_logis[0]:.4f}")
print(f"Koefisien (theta_1, theta_2): {theta_logis[1:]}")

# Plot Log Loss
plt.figure(figsize=(8, 4))
plt.plot(loss_hist, color='purple')
plt.title('Kurva Log Loss selama Pelatihan')
plt.xlabel('Iterasi')
plt.ylabel('Loss (Binary Cross-Entropy)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Soal B3 – Evaluasi Model

```python
X_test_lb = np.c_[np.ones((X_test_l.shape[0], 1)), X_test_l]
y_test_prob = sigmoid(X_test_lb.dot(theta_logis))
y_test_pred = (y_test_prob >= 0.5).astype(int)

# Confusion Matrix
tp = np.sum((y_test_l == 1) & (y_test_pred == 1))
tn = np.sum((y_test_l == 0) & (y_test_pred == 0))
fp = np.sum((y_test_l == 0) & (y_test_pred == 1))
fn = np.sum((y_test_l == 1) & (y_test_pred == 0))

accuracy = (tp + tn) / len(y_test_l)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print("Confusion Matrix:")
print(f"  TN: {tn} | FP: {fp}")
print(f"  FN: {fn} | TP: {tp}")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
```

### Soal B4 – Decision Boundary

```python
# Visualisasi Decision Boundary
plt.figure(figsize=(8, 5))
plt.scatter(X_test_l[y_test_l == 0, 0], X_test_l[y_test_l == 0, 1], color='blue', label='Bukan Spam (Aktual 0)', edgecolor='k', s=80)
plt.scatter(X_test_l[y_test_l == 1, 0], X_test_l[y_test_l == 1, 1], color='red', label='Spam (Aktual 1)', edgecolor='k', s=80)

x_bd = np.linspace(0, 10, 100)
y_bd = -(theta_logis[0] + theta_logis[1] * x_bd) / theta_logis[2]

plt.plot(x_bd, y_bd, color='green', linewidth=2, label='Decision Boundary (P=0.5)')
plt.xlim(-1, 11)
plt.ylim(-1, 11)
plt.title('Decision Boundary Logistic Regression')
plt.xlabel('Jumlah Link')
plt.ylabel('Kata Promo')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Soal B5 – Eksperimen Threshold

```python
thresholds = [0.3, 0.5, 0.7]
for th in thresholds:
    pred_th = (y_test_prob >= th).astype(int)
    tp_t = np.sum((y_test_l == 1) & (pred_th == 1))
    tn_t = np.sum((y_test_l == 0) & (pred_th == 0))
    fp_t = np.sum((y_test_l == 0) & (pred_th == 1))
    fn_t = np.sum((y_test_l == 1) & (pred_th == 0))
    
    acc_t = (tp_t + tn_t) / len(y_test_l)
    prec_t = tp_t / (tp_t + fp_t) if (tp_t + fp_t) > 0 else 0
    rec_t = tp_t / (tp_t + fn_t) if (tp_t + fn_t) > 0 else 0
    f1_t = 2 * (prec_t * rec_t) / (prec_t + rec_t) if (prec_t + rec_t) > 0 else 0
    
    print(f"Threshold {th}: Acc={acc_t:.4f}, Prec={prec_t:.4f}, Rec={rec_t:.4f}, F1={f1_t:.4f}, CM=[[{tn_t}, {fp_t}], [{fn_t}, {tp_t}]]")
```

### Soal B6 – Prediksi Email Baru (Threshold = 0.5)

```python
email_baru = np.array([
    [1, 2],
    [5, 6],
    [9, 10],
    [2, 11]
])

email_baru_b = np.c_[np.ones((len(email_baru), 1)), email_baru]
prob_baru = sigmoid(email_baru_b.dot(theta_logis))
pred_baru = (prob_baru >= 0.5).astype(int)

for i, (e, p, pr) in enumerate(zip(email_baru, prob_baru, pred_baru), 1):
    status = "1 (Spam)" if pr == 1 else "0 (Bukan Spam)"
    print(f"Email E{i} (Link={e[0]}, Promo={e[1]}): Prob_Spam = {p:.4f} -> Pred_Spam = {status}")
```

## Bagian C – Pemahaman Konsep (Teori)

* **Jawaban C1**: Logistic Regression dinamakan "regression" karena basis matematikanya menggunakan model regresi linier untuk menghitung log-odds dari target biner. **Fungsi sigmoid** berperan mengubah output kontinu dari regresi linier tersebut menjadi nilai probabilitas antara $0$ dan $1$.
* **Jawaban C2**: Normal equation menghitung solusi analitik secara langsung menggunakan matriks invers (sangat cepat untuk dataset kecil, namun lambat $O(n^3)$ untuk fitur besar). Gradient descent adalah metode optimasi iteratif berbasis turunan yang sangat ideal digunakan ketika ukuran dataset sangat besar.
* **Jawaban C3**: Evaluasi pada data testing wajib dilakukan untuk mengukur kemampuan generalisasi model terhadap data baru yang belum pernah dilihat saat pelatihan. **Overfitting** adalah kondisi ketika model terlalu menghafal data training sehingga gagal berkinerja baik pada data testing.
* **Jawaban C4**: Linear Regression menggunakan MSE karena targetnya berupa nilai kontinu dengan asumsi distribusi normal. Sebaliknya, Logistic Regression menggunakan log loss (binary cross-entropy) karena outputnya berupa probabilitas diskrit, di mana MSE akan menghasilkan fungsi cost non-konveks yang sulit dioptimalkan.
* **Jawaban C5**: Perbedaan angka bisa terjadi karena faktor **random_state** (pembagian data train-test yang berbeda jika tidak diset sama), inisialisasi bobot awal pada gradient descent, atau perbedaan pustaka/versi library yang digunakan.
